**Project: Credit Risk Prediction — Machine Learning-Based Credit Risk Assessment for Default Prediction in Consumer Lending**

This module loads the raw dataset and takes a first look at it: how many rows and columns, whether anything is missing, and what the target column (`Default`) looks like.

**Dataset:** `Loan_default.csv` — 255,347 real loan applications. Each row is one borrower.
`Default` = 1 means the borrower did not pay back the loan, 0 means they did.

**Input:** `data/Loan_default.csv`
**Output:** printed overview only (no files saved by this module)


```python
import os
import pandas as pd

DATA_PATH = "data/Loan_default.csv"
TARGET = "Default"

pd.set_option("display.width", 120)
pd.set_option("display.max_columns", 50)

df = pd.read_csv(DATA_PATH)

print(f"Rows (borrowers)  : {df.shape[0]:,}")
print(f"Columns           : {df.shape[1]}")
print("\nFirst 5 rows:")
df.head()
```

```python
print("Column names and their types:")
print(df.dtypes)

print("\nAny missing values?")
missing = df.isnull().sum()
missing = missing[missing > 0]
if len(missing):
    print(missing)
else:
    print("None - every cell in the dataset has a value.")

print(f"\nHow many borrowers defaulted?")
print(df[TARGET].value_counts())
print((df[TARGET].value_counts(normalize=True) * 100).round(2).astype(str) + "%")
```

## How to read this

- We have **255,347 borrowers** and **18 columns** — plenty of data.
- **No missing values anywhere.** Real datasets usually have some, so this is lucky.
- Only **11.6% of borrowers defaulted** (about 1 in 9). Keep this figure in mind — it explains a lot of what comes later. A model that always guesses "no default" would already be 88% "accurate" without being useful, so accuracy alone will be misleading when we score models in Module 6.
- `LoanID` is a random ID per borrower and carries no risk information. It gets dropped in Module 3 (Data Preprocessing).
